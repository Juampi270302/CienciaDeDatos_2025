```python
import plotly.express as px
import importlib
import plotly.io as pio
import pandas as pd

def dibujar_histograma(df, atributo):
    fig = px.histogram(df, x=atributo, nbins=50, title=f'Distribución de {atributo}')
    fig.update_layout(
        paper_bgcolor='black',
        plot_bgcolor='black',
        font_color='white'
    )
    #fig.show() # Descomentar para ver los graficos en el notebook
    pio.write_html(fig, file=f"./histogramas_post_etl/histograma_{atributo}.html", auto_open=False, full_html=True, include_plotlyjs='cdn')
```

```python
df_etl = pd.read_csv('../dataset/datos_limpios.csv', skipinitialspace=True)
df_etl
```

```python
dibujar_histograma(df_etl, "average_token_length")
```
